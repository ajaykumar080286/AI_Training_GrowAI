***Measure LLM Processing TIme***

```python
import time
def measure_time(func):
  def wrapper(*args,**kwargs):
    start_time=time.time()
    result=func(*args,**kwargs)
    end_time=time.time()
    execution_time=end_time-start_time
    print(f"Execution time:", round(end_time-start_time,3),"seconds")
    return result
  return wrapper

```

```python
@measure_time
def process_llm_execution(question):
  print("processing",question)
  time.sleep(1)
  return "LLM response"

resposne=process_llm_execution("whay is llm")
print(resposne)
```

**Cosine Similarity**

Similarity between two vector

```python
embedding_a=np.array([1,0,1])
embedding_b=np.array([1,1,0])
```

```python
dot_product=np.dot(embedding_a,embedding_b)
norm_a=np.linalg.norm(embedding_a)
norm_b=np.linalg.norm(embedding_b)

cosine_similiryt=dot_product/(norm_a*norm_b)
print(cosine_similiryt)
```

```python
from pydantic import BaseModel, Field

class LLMRequest(BaseModel):
    question: str
    model_name: str = "llama3"
    temperature: float = Field(default=0.7, ge=0, le=2)
    max_tokens: int = Field(default=500, gt=0)

#Object of Class
request = LLMRequest(
    question="Explain RAG",
    model_name="llama3",
    temperature=0.7,
    max_tokens=500
)

print(request)
```
