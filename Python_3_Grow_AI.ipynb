***How to Create YAML Configuration file***

Create a file name: llm_config_file

```python
yaml_config = """
application: AI_assistent
model:
    name: gpt4
    tempertaure: 1.0
    max_token: 1000
system:
   rule: python tutor
   Language: English
feature:
   stream: True
   conversion_memory: true
"""

print("YAML configuration string stored successfully.")
```

**read YML using python**

```python
import yaml

# The configuration we want to write

# 1. First, save the YAML string into a file
with open("llm_config.yaml", "w") as file:
    file.write(yaml_config)

# 2. Now, read and parse the YAML file
with open("llm_config.yaml", "r") as file:
    config = yaml.safe_load(file)

print(config)
```
